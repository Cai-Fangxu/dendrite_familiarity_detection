### Load grid search data from .csv files to dataframe and write dataframe to a tmp file, to be processed by `process_gridSearch_df.ipynb`. 

```python
import numpy as np
import pandas as pd
import scipy.stats
```

```python
ndRlist = [1, 2, 3, 4, 5, 10, 12, 15, 20, 30, 60, 100, 300]
ndlist = [1, 2, 3, 4, 5, 10, 12, 15, 20, 30, 50, 60, 75, 100, 150, 300, 400, 750, 1000, 1500]
nslist = [60000, 30000, 20000, 15000, 12000, 6000, 5000, 4000, 3000, 2000, 1200, 1000, 800, 600, 400, 200, 150, 80, 60, 40]
```

```python
# file_path0 = "../data/no_nonlinearity/"
# file_path0 = "../data/gridSearch_standard/60000/coeff_infty/"
file_path0 = "../data/gridSearch_SomaNoise/coeff_infty/dendrite_noise_0.15/"
# file_path0 = "../tmp_data/coeff_infty/dendrite_noise_0.25_soma_noise_0.1/"

0.5, .2, .25, .3
```

```python
n_trials  = 6

df = pd.DataFrame(data=None)
for nd_idx, nd in enumerate(ndlist):
    for i, ndR in enumerate(ndRlist):
        tmp_df = pd.DataFrame(data=None)
        try:
            ns=nslist[nd_idx]
            tmp_array = np.genfromtxt(f"{file_path0}nd{nd}_ns{ns}/n{ndR}.csv", delimiter=',')
            n_rows = tmp_array.shape[0]
            tmp_df.insert(0, "nd", np.ones(n_rows)*nd)
            tmp_df.insert(1, "ndR", np.ones(n_rows)*ndR)
            tmp_df.insert(2, "b", tmp_array[:, 0])
            tmp_df.insert(3, "bk", tmp_array[:, 1])
            tmp_df.insert(4, "beta", tmp_array[:, 2])
            tmp_array_mean = np.mean(tmp_array[:, -n_trials:], axis=1)
            tmp_array_sem = scipy.stats.sem(tmp_array[:, -n_trials:], axis=1)
            tmp_df.insert(5, "capacity", tmp_array_mean)
            tmp_df.insert(6, "sem", tmp_array_sem)
            df = pd.concat([df, tmp_df], ignore_index=True)
        except:
            pass
```

```python
df.to_csv("../tmp_data/plot_data/tmp_df.csv", index=False)
```
